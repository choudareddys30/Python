```python
l = [1,2,3,4,5,6,7,8,9,3,5,3,6,7,8,10,53]
len(l)
for i in range(0,len(l),2):
  if l[i] % 2 == 0:
    print(f"{l[i]} is a Even Number")
  else:
    print(f"{l[i]} is a Odd Number")
```
