```python
number  = int(input("Enter a Temperature"))
if number > 0:
  print(f"Temperature is greater than freezing point: {number}")
elif number == 0:
  print(f"Temperature is at freezing Point: {number}")
else:
  print(f"Temperature is below freezing point: {number}")
```
