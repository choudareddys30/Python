```python
def notification(**kwargs):
  if "name" in kwargs and "email" in kwargs:
    return f"Notification sent to {kwargs['name']} at {kwargs['email']}"
  elif "name" in kwargs:
    return f"Notification sent to {kwargs['name']}"
  elif "email" in kwargs:
    return f"Notification sent to {kwargs['email']}"
  else:
    return f"Missing required Information"
```

```python
def factorial(n):
  if n == 0  or n == 1:
    return 1
  else:
    return n * factorial(n-1)
```

```python
factorial(3)
```

```python
factorial(4)
```
